```python
%%R
library(ggplot2)
library(dplyr)

train <- read.csv("/kaggle/input/titanicdataset-traincsv/train.csv")

survival_rate <- train %>%
  group_by(Sex) %>%
  summarise(SurvivalRate = mean(Survived, na.rm = TRUE))

print(survival_rate)

ggplot(train, aes(x = Sex, fill = factor(Survived))) +
  geom_bar(position = "fill") +
  labs(y = "Proportion", fill = "Survived") +
  ggtitle("Survival Rate by Sex in Titanic Dataset") +
  theme_minimal()
```
